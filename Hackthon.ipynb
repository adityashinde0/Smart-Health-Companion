# Smart Health Companion - DeepSeek-V3 Integration Notebook
This notebook demonstrates using DeepSeek-V3 via CloudRift's OpenAI-compatible inference endpoint for health recommendations.

```python
import os
from openai import OpenAI

# Securely fetch API key from environment variable
api_key = os.getenv("OPENAI_API_KEY", "YOUR_API_KEY_HERE")

client = OpenAI(
    api_key=api_key,
    base_url="https://inference.cloudrift.ai/v1"
)

completion = client.chat.completions.create(
    model="deepseek-ai/DeepSeek-V3",
    messages=[
        {"role": "user", "content": "Explain how machine learning assists in cardiovascular disease risk assessment."}
    ],
    stream=True
)
for chunk in completion:
    if chunk.choices and len(chunk.choices) > 0:
        delta = chunk.choices[0].delta
        if delta and delta.content:
            print(delta.content, end="")
```

```python
def get_health_advice(symptom: str):
    """Generate health advice using DeepSeek-V3 with a clinical disclaimer."""
    prompt = f"You are a smart health companion. A user reports: '{symptom}'. Provide an educational summary, risk considerations, and questions for their physician."
    response = client.chat.completions.create(
        model="deepseek-ai/DeepSeek-V3",
        messages=[{"role": "user", "content": prompt}],
        stream=True
    )
    for chunk in response:
        if chunk.choices and chunk.choices[0].delta.content:
            print(chunk.choices[0].delta.content, end="")
    print("\n\nNote: This advice is AI-generated for educational purposes. Always consult a licensed healthcare professional for diagnosis.")
```
